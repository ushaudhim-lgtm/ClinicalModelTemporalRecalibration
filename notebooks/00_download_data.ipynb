This notebook downloads the NHAMCS Emergency Department data files and the codebook pdfs
for years 2011 to 2019 from the CDC website. It saves the zip files and pdfs into the
data/raw and data/metadata folders, then unzips the data files so they are ready to use
in the next notebooks. Only needs to be run once, it skips files that are already there.

```python
import sys
sys.path.append(".")
import os
import zipfile
from utils import STUDY_YEARS, DATA_FILES, DOC_FILES, DATA_BASE_URL, DOC_BASE_URL, RAW_DIR, META_DIR, download_file, raw_file_path

os.makedirs(RAW_DIR, exist_ok=True)
os.makedirs(META_DIR, exist_ok=True)
```

```python
# download the zip data files and the pdf codebooks for every year
for yr in STUDY_YEARS:
    data_url = DATA_BASE_URL + "/" + DATA_FILES[yr]
    zip_path = os.path.join(RAW_DIR, DATA_FILES[yr])
    download_file(data_url, zip_path)

    doc_url = DOC_BASE_URL + "/" + DOC_FILES[yr]
    doc_path = os.path.join(META_DIR, DOC_FILES[yr])
    download_file(doc_url, doc_path)

print("done downloading")
```

```python
# unzip every year's data file, cdc zips have inconsistent casing inside so fix that up too
for yr in STUDY_YEARS:
    zip_path = os.path.join(RAW_DIR, DATA_FILES[yr])
    out_path = raw_file_path(yr)
    if os.path.exists(out_path):
        print(yr, "already unzipped")
        continue
    with zipfile.ZipFile(zip_path) as z:
        inner_name = z.namelist()[0]
        z.extract(inner_name, RAW_DIR)
    extracted_path = os.path.join(RAW_DIR, inner_name)
    if extracted_path != out_path:
        os.rename(extracted_path, out_path)
    print(yr, "unzipped ->", out_path)
```
